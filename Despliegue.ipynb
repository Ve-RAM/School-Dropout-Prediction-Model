```python
import joblib
import numpy as np

# Cargar el modelo entrenado desde el archivo .pkl
modelo_path = 'modelo_final_escogido.pickle'  # Asegúrate de que el archivo esté en el mismo directorio
modelo = joblib.load(modelo_path)

# Lista de características necesarias para realizar la predicción (ajusta los nombres a tus columnas)
columnas = ['AÑO', 'CÓDIGO_MUNICIPIO', 'CÓDIGO_ETC', 'POBLACIÓN_5_16', 'TASA_MATRICULACIÓN_5_16',
       'COBERTURA_NETA_SECUNDARIA', 'COBERTURA_BRUTA_SECUNDARIA',
       'APROBACIÓN_SECUNDARIA', 'REPROBACIÓN_SECUNDARIA',
       'REPITENCIA_SECUNDARIA']

# Función para pedir datos al usuario
def obtener_datos_usuario(columnas):
    print("Por favor, ingrese los valores para las siguientes características:")
    datos = []
    for columna in columnas:
        while True:
            try:
                valor = float(input(f"Ingrese el valor para {columna}: "))
                datos.append(valor)
                break
            except ValueError:
                print("Por favor, ingrese un número válido.")
    return np.array(datos).reshape(1, -1)

# Obtener los datos del usuario
registro_usuario = obtener_datos_usuario(columnas)

# Realizar la predicción
prediccion = modelo.predict(registro_usuario)

# Mostrar el resultado al usuario
print(f"\nLa predicción de 'DESERCIÓN_SECUNDARIA' para el registro ingresado es: {prediccion[0]}")
```
